````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv("russian-housing-market.csv")
print(df)
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
df_empty_life_sq = df[df["life_sq"].isnull()]
count_empty_life_sq = len(df_empty_life_sq)
print(count_empty_life_sq)
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
ratio = (df["life_sq"] / df["full_sq"]).median()
result_df = df.copy()
result_df["life_sq"] = df["life_sq"].fillna(df["full_sq"] * ratio)
print(result_df)

print(
    "жилая площадь сильно коррелирует с полной, а медианное отношение устойчиво к выбросам. Замена через среднее или медиану life_sq проигрывает: она даст одинаковую жилую площадь для студии и для четырёхкомнатной квартиры")
````

3. Какие значения имеет 25, 50 и 75 перцентили полной площади помещения

````python
result = df["full_sq"].quantile([0.25, 0.5, 0.75])
print(result)
````

4. Посчитайте среднюю стоимость продажи помещений с общей площадью от 50 до 70 квадратных метров

````python
filter_df = df[(df["full_sq"] <= 70) & (df["full_sq"] >= 50)]
result = filter_df["price_doc"].mean()
print(result)
````

5. Какая максимальная этажность здания представлена в датасете

````python
# выбираю специально между floor и max_floor, потому что max_floor может быть nan, в таком случае нужно брать из floor
result = df[["floor", "max_floor"]].max().max()
print(result)
````

6. Выявите район с самым дорогим жильем

````python
# самое дорогое жилье - не там где самая дорогая сделка, а там где самая большая медианная сделка
area_prices = df.groupby("sub_area")["price_doc"].median()
most_expensive_area = area_prices.sort_values(ascending=False).head(n=1)
print(most_expensive_area)
````

7. Посчитайте среднюю стоимость за квадратный метр жилья в Митино(Mitino)

````python
mitino_flats = df[df["sub_area"] == "Mitino"]
mitino_price_per_square = mitino_flats["price_doc"] / mitino_flats["full_sq"]
result = mitino_price_per_square.mean()
print(result)
````

8. Посчитайте среднюю площадь комнаты в Бибирево

````python
# life_sq, потому что общая площадь включает балконы, лоджии и нежилые помещения
bibirevo_flats = df[(df["sub_area"] == "Bibirevo") & (df["num_room"].notna())]
bibirevo_room_square = bibirevo_flats["life_sq"] / bibirevo_flats["num_room"]
result = bibirevo_room_square.mean()
print(result)
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

````python
flats_1_floor = df[df["floor"] == 1].copy()
flats_9_floor = df[df["floor"] == 9].copy()

flats_1_floor["price_per_square"] = flats_1_floor["price_doc"] / flats_1_floor["full_sq"]
flats_9_floor["price_per_square"] = flats_9_floor["price_doc"] / flats_9_floor["full_sq"]

median_price_1_floor = flats_1_floor["price_doc"].median()
median_price_9_floor = flats_9_floor["price_doc"].median()

median_price_per_square_1_floor = flats_1_floor["price_per_square"].median()
median_price_per_square_9_floor = flats_9_floor["price_per_square"].median()

print(
    f"Медианная цена за квартиры на первом этаже: {median_price_1_floor}, Медианная цена за квартиры на девятом этаже: {median_price_9_floor}")
print(
    f"Медианная цена за квадратный метр квартиры на первом этаже: {median_price_per_square_1_floor}, Медианная цена за квадратный метр квартиры на девятом этаже: {median_price_per_square_9_floor}")

diff_1_9_floors = median_price_9_floor - median_price_1_floor
diff_per_square_1_9_floors = median_price_per_square_9_floor - median_price_per_square_1_floor
print(f"Разница в цене между квартирами на 9 и на 1 этаже: {diff_1_9_floors} рублей")
print(f"Разница в цене между квадратным метром квартиры на 9 и на 1 этаже: {diff_per_square_1_9_floors} рублей")
print(
    "Квартиры на 9 этаже дороже по медианной цене на ~830 тыс. рублей, но цена за квадратный метр отличается лишь на ~4 тыс. рублей (3%). Значит, основная разница в общей цене объясняется большей площадью квартир, а не самим этажом.")
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

df["timestamp"] = pd.to_datetime(df["timestamp"])
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
area_nagatinski = df[df["sub_area"] == "Nagatinskij Zaton"]

monthly = area_nagatinski.groupby(area_nagatinski["timestamp"].dt.to_period("M"))["price_doc"].agg(["mean", "count"])
monthly.index = monthly.index.to_timestamp()

fig, ax = plt.subplots(figsize=(11, 4.5))
# столбики: число сделок в месяце — показывают, где среднее построено по 1-2 сделкам
ax2 = ax.twinx()
ax2.bar(monthly.index, monthly["count"], width=20, color="lightgray", alpha=0.7)
ax2.set_ylabel("Число сделок в месяц")
ax.set_zorder(ax2.get_zorder() + 1)
ax.patch.set_visible(False)

ax.plot(monthly.index, monthly["mean"] / 1_000_000, marker="o", color="tab:blue")
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
fig.autofmt_xdate()
ax.set_xlabel("Время")
ax.set_ylabel("Средняя цена, млн рублей")
ax.set_title("Средняя цена квартиры в Nagatinskij Zaton по месяцам")
ax.grid(alpha=0.3)
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
prices = df[df["full_sq"] > 50]["price_doc"] / 1_000_000

# выбросы (до 111 млн) растягивают ось, поэтому показываем до 99-го персентиля
upper = prices.quantile(0.99)
shown = prices[prices <= upper]

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(shown, bins=40, edgecolor="white")
ax.axvline(prices.median(), color="tab:red", linestyle="--", label=f"Медиана: {prices.median():.1f} млн")
ax.set_xlabel("Цена, млн рублей")
ax.set_ylabel("Количество квартир")
ax.set_title(f"Распределение цен квартир с площадью > 50 м² (до 99-го перцентиля, {upper:.0f} млн)")
ax.legend()
plt.show()
print(f"Не показано выбросов выше {upper:.1f} млн: {(prices > upper).sum()} из {len(prices)}")
````

3. Визуализируйте количество проданных квартир в разных районах

````python
TOP_N = 20
sales = df.groupby("sub_area").size().sort_values(ascending=False).head(TOP_N).sort_values()

fig, ax = plt.subplots(figsize=(9, 7))
ax.barh(sales.index, sales.values)
ax.set_xlabel("Количество проданных квартир")
ax.set_title(f"Топ-{TOP_N} районов по числу сделок (из {df['sub_area'].nunique()})")
for y, v in enumerate(sales.values):
    ax.text(v, y, f" {v}", va="center")
plt.tight_layout()
plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
basmanny = df[df["sub_area"] == "Basmannoe"]["full_sq"].dropna()

fig, ax = plt.subplots(figsize=(4, 6))
ax.boxplot(basmanny)
ax.set_xticklabels(["Basmannoe"])
ax.set_ylabel("Полная площадь, квадратный метр")
ax.set_title("Распределение полных площадей, Басманный район")
plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
g = df.groupby("floor")["price_doc"].agg(["mean", "count"])
g = g[g["count"] >= 20]  # отбрасываем этажи с малым числом наблюдений

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(g.index, g["mean"] / 1_000_000, marker="o")
ax.set_xlabel("Этаж")
ax.set_ylabel("Средняя цена, млн рублей")
ax.set_title("Средняя цена квартиры в зависимости от этажа (этажи с ≥ 20 сделок)")
ax.set_xticks(g.index.astype(int))
ax.grid(alpha=0.3)
plt.show()
````
